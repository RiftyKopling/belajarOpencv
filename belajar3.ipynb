## <font style="color:black">Menggambar sebuah garis</font>

mari mulai dari menggambar garis di dalam sebuah gambar, kita akan memakai fungsi cv2.line untuk ini.

### <font style="color:rgb(8,133,37)">intaks fungsi cv2.line</font>
``` python
    img = cv2.line(img, pt1, pt2, color[, thickness[, lineType[, shift]]])
```

`img`: output gambar yang di beri garis.

fungsi ini memerlukan **4 arguments wajib**:

1. `img`:   gambar mana yang akan kita beri garis

2. `pt1`:   titik pertama(x,y location) dari garis yang kita gambar

3. `pt2`:   point kedua dari garis 

4. `color`: warna dari garis yang kita gambar
    
selain argument wajib ada juga argument optional yang perlu diketahui:

1. `thickness`: Integer untuk mengatur ketebalan garis. Default nya adalah 1.

2. `lineType`:  Jenis garis. Nilai defaultnya adalah 8, yang berarti garis dengan konektivitas 8 (8-connected line). Biasanya, cv2.LINE_AA (garis anti-aliasing atau halus) digunakan untuk lineType.
    
### <font style="color:rgb(8,133,37)">Dokumentasi OpenCV</font>

**`line`**: <a href="https://docs.opencv.org/4.5.1/d6/d6e/group__imgproc__draw.html#ga7078a9fae8c7e7d13d24dac2520ae4a2" target="_blank">link dokumentasi</a>


````python
gambar_garis = bahan.copy()

cv2.line(gambar_garis, (100, 500), (500, 500), (0, 0, 255), 2)

plt.imshow(gambar_garis[:,:,::-1]), plt.title(" Membuat garis")
````

````python
plt.figure(figsize=(40, 20))
plt.subplot(1,4,1);plt.imshow(cv2.line(gambar_garis, (100, 500), (500, 500), (0, 0, 255), 5, cv2.LINE_AA)), plt.title(" Membuat garis dengan LINE_AA")
plt.subplot(1,4,2);plt.imshow(cv2.line(gambar_garis, (100, 500), (500, 500), (0, 0, 255), 5, cv2.LINE_8)), plt.title(" Membuat garis dengan LINE_8")
````

# Menggambar Lingkaran

Mari kita mulai dengan menggambar lingkaran pada sebuah gambar. Kita akan menggunakan fungsi `cv2.circle` untuk ini.

### Sintaks Fungsional

```python
img = cv2.circle(img, center, radius, color[, thickness[, lineType[, shift]]])

```

`img`: Gambar keluaran yang telah dianotasi.

Fungsi ini memiliki **4 argumen wajib**:

1. `img`: Gambar di mana kita akan menggambar garis
2. `center`: Titik pusat lingkaran
3. `radius`: Jari-jari lingkaran
4. `color`: Warna lingkaran yang akan digambar

Selanjutnya, mari kita lihat argumen (opsional) yang akan cukup sering kita gunakan.

1. `thickness`: Ketebalan garis tepi lingkaran (jika bernilai positif).
Jika nilai negatif diberikan untuk argumen ini, hasilnya akan berupa lingkaran yang terisi penuh.
2. `lineType`: Jenis batas lingkaran. Ini sama persis dengan argumen `lineType` pada **cv2.line**

### Dokumentasi OpenCV

**`circle`**:  <a href="https://docs.opencv.org/4.5.1/d6/d6e/group__imgproc__draw.html#gaf10604b069374903dbd0f0488cb43670" target="_blank">Documentation link</a>

Mari kita lihat contoh dari hal ini.

````python
gambar_lingkaran = bahan.copy()

cv2.circle(gambar_lingkaran, (500,400), 250, (0, 0, 255), -5, cv2.LINE_AA)

plt.imshow(gambar_lingkaran), plt.title(" Membuat Lingkaran")
````

# Menggambar Persegi Panjang
Kita akan menggunakan fungsi cv2.rectangle untuk menggambar persegi panjang pada sebuah gambar. Sintaks fungsinya adalah sebagai berikut.

### Sintaks Fungsional
```python
img = cv2.rectangle(img, pt1, pt2, color[, thickness[, lineType[, shift]]])
```

`img` : Gambar keluaran yang telah dianotasi.

Fungsi ini memiliki 4 argumen wajib:

1. `img`: Gambar di mana persegi panjang akan digambar.

2. `pt1`: Titik sudut persegi panjang. Biasanya kita menggunakan titik sudut kiri atas di sini.

3. `pt2`: Titik sudut persegi panjang yang berlawanan dengan pt1. Biasanya kita menggunakan titik sudut kanan bawah di sini.

4. `color`: Warna persegi panjang

Selanjutnya, mari kita lihat argumen (opsional) yang akan cukup sering kita gunakan.

1. `thickness`: Ketebalan garis tepi persegi panjang (jika bernilai positif). Jika nilai negatif diberikan untuk argumen ini, hasilnya akan berupa persegi panjang yang terisi penuh.

2. `lineType`: Jenis batas persegi panjang. Ini sama persis dengan argumen lineType pada cv2.line

Tautan Dokumentasi OpenCV
rectangle: <a href="https://docs.opencv.org/4.5.1/d6/d6e/group__imgproc__draw.html#ga07d2f74cadcf8e305e810ce8eed13bc9" target="_blank">Documentation link</a>

Mari kita lihat contoh dari hal ini.

````python
gambar_kotak = bahan.copy()

cv2.rectangle(gambar_kotak, (200,200), (1000, 500), (255, 0, 0), 8, cv2.LINE_AA)

plt.imshow(gambar_kotak), plt.title("Membuat Gambar Kotak")
````

## <font style="color:black">Menambahkan Teks</font>

Terakhir, mari kita lihat bagaimana cara menulis teks pada gambar menggunakan fungsi **cv2.putText**.

### <font style="color:rgb(8,133,37)">Sintaks Fungsional</font>

```python
img = cv2.putText(img, text, org, fontFace, fontScale, color[, thickness[, lineType[, bottomLeftOrigin]]])
```

`img`: Gambar keluaran yang telah dianotasi.

Fungsi ini memiliki **6 argumen wajib**:

1. `img`: Gambar di mana teks akan ditulis.

2. `text`: String teks yang akan ditulis.

3. `org`: Sudut kiri bawah dari string teks pada gambar.

4. `fontFace`: Jenis font.

5. `fontScale`: Faktor skala font yang dikalikan dengan ukuran dasar spesifik font.

6. `color`: Warna font.

Argumen opsional lainnya yang penting untuk kita ketahui meliputi:

1. `thickness`: Bilangan bulat yang menentukan ketebalan garis untuk teks. Nilai bawaannya adalah 1.

2. `lineType`: Jenis garis. Nilai bawaannya adalah 8 yang berarti *8-connected line*. Biasanya, `cv2.LINE_AA` (*antialiased* atau garis halus) digunakan untuk `lineType`.

### <font style="color:rgb(8,133,37)">Dokumentasi OpenCV</font>

**`putText`**: <a href="https://docs.opencv.org/4.5.1/d6/d6e/group__imgproc__draw.html#ga5126f47f883d730f633d74f07456c576" target="_blank">Tautan Dokumentasi</a>
    
Mari kita lihat contoh dari hal ini.

````python
gambar_tulisan = bahan.copy()
fontScale = 3
fontFace = cv2.FONT_HERSHEY_PLAIN
fontColor = (255, 0, 0)
fontThickness = 2

cv2.putText(gambar_tulisan, "Gambar Roket Perodok Planet Mars", (200,100), fontFace, fontScale, fontColor, fontThickness)

plt.imshow(gambar_tulisan), plt.title("Membuat Tulisan")
````

````python
gambar_tulisan_kebalik = bahan.copy()
fontScale = -1
fontFace = cv2.FONT_HERSHEY_PLAIN
fontColor = (255, 0, 0)
fontThickness = 2

cv2.putText(gambar_tulisan_kebalik, "Gambar Roket Perodok Planet Mars", (200,100), fontFace, fontScale, fontColor, fontThickness)

plt.imshow(gambar_tulisan_kebalik), plt.title("Membuat Tulisan Kebalik")
````
